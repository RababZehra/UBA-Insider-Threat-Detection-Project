```python
# -*- coding: utf-8 -*-

!pip -q install torch-geometric

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.nn.functional as F

from torch_geometric.data import Data
from torch_geometric.nn import GATConv

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

from tqdm.auto import trange
from google.colab import drive

drive.mount('/content/drive')

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print("Device:", device)


# -----------------------------------
# 1. PATHS AND DIRECTORIES
# -----------------------------------

BASE_DIR = "/content/drive/MyDrive/Insider_Threat_Project"

FEATURE_PATH = os.path.join(
    BASE_DIR,
    "lstm_feature_extraction_result",
    "1-data-test-manual+lstm_seq74.csv"
)

EDGE_PATH = os.path.join(
    BASE_DIR,
    "sample_data",
    "1-data-test-undirected_edge.csv"
)

RESULT_DIR = os.path.join(
    BASE_DIR,
    "GAT_results"
)

MODEL_DIR = os.path.join(
    RESULT_DIR,
    "early_stop_model"
)

os.makedirs(RESULT_DIR, exist_ok=True)
os.makedirs(MODEL_DIR, exist_ok=True)


# -----------------------------------
# 2. LOAD DATA
# -----------------------------------

df_data = pd.read_csv(FEATURE_PATH)
df_e = pd.read_csv(EDGE_PATH)

print("Feature data shape:", df_data.shape)
print("Edge data shape:", df_e.shape)

assert "label" in df_data.columns, "Label column is missing."
assert all(col in df_e.columns for col in ["start", "end"]), \
    "Edge file must contain start and end columns."

cols = df_data.columns[5:]

x = torch.tensor(
    df_data[cols].to_numpy(dtype=np.float32),
    dtype=torch.float32
)

y = torch.tensor(
    df_data["label"].to_numpy(dtype=np.int64),
    dtype=torch.long
)

edge_index = torch.tensor(
    df_e[["start", "end"]].to_numpy(dtype=np.int64).T,
    dtype=torch.long
)


# -----------------------------------
# 3. DATA VALIDATION
# -----------------------------------

assert x.shape == (1908, 10), \
    f"Unexpected feature shape: {x.shape}"

assert y.shape == (1908,), \
    f"Unexpected label shape: {y.shape}"

assert edge_index.shape[0] == 2, \
    "Edge index must have two rows."

assert edge_index.shape[1] > 0, \
    "No edges found in the edge file."

assert edge_index.min().item() >= 0, \
    "Negative node index found."

assert edge_index.max().item() < len(df_data), \
    "Edge index exceeds the number of nodes."

assert torch.isfinite(x).all(), \
    "Input features contain NaN or infinite values."

assert set(torch.unique(y).tolist()).issubset({0, 1}), \
    "This code expects binary labels 0 and 1."

print("\nSelected feature columns:", list(cols))
print("Input feature shape:", tuple(x.shape))
print("Label shape:", tuple(y.shape))
print("Edge index shape:", tuple(edge_index.shape))


# -----------------------------------
# 4. CREATE GRAPH DATA
# -----------------------------------

data = Data(
    x=x,
    edge_index=edge_index,
    y=y
)

print("\nGraph data:")
print(data)

data.train_mask = torch.zeros(
    data.num_nodes,
    dtype=torch.bool
)

data.test_mask = torch.zeros(
    data.num_nodes,
    dtype=torch.bool
)

data.train_mask[:data.num_nodes - 572] = True
data.test_mask[data.num_nodes - 572:] = True

print("\nNumber of nodes:", data.num_nodes)
print("Number of node features:", data.num_node_features)
print("Number of edges:", data.edge_index.shape[1])
print("Training nodes:", int(data.train_mask.sum()))
print("Testing nodes:", int(data.test_mask.sum()))

data = data.to(device)


# -----------------------------------
# 5. GAT MODEL
# -----------------------------------

class GATNet(nn.Module):

    def __init__(
        self,
        input_dim,
        hidden_dim=16,
        num_classes=2,
        heads=4,
        dropout=0.5
    ):
        super(GATNet, self).__init__()

        self.dropout = dropout

        self.gat1 = GATConv(
            input_dim,
            hidden_dim,
            heads=heads,
            dropout=dropout
        )

        self.gat2 = GATConv(
            hidden_dim * heads,
            num_classes,
            heads=1,
            concat=False,
            dropout=dropout
        )

    def forward(self, data):

        x, edge_index = data.x, data.edge_index

        x = self.gat1(x, edge_index)

        x = F.elu(x)

        x = F.dropout(
            x,
            p=self.dropout,
            training=self.training
        )

        x = self.gat2(x, edge_index)

        return F.log_softmax(x, dim=1)


# -----------------------------------
# 6. EARLY STOPPING
# -----------------------------------

class EarlyStopping:

    def __init__(
        self,
        save_path,
        patience=30,
        verbose=True,
        delta=0.0001,
        metric='loss'
    ):

        self.save_path = save_path
        self.patience = patience
        self.verbose = verbose
        self.counter = 0
        self.best_score = None
        self.early_stop = False
        self.delta = delta
        self.metric = metric
        self.val_bound = np.inf
        self.sign = -1 if metric == 'loss' else 1

    def __call__(self, value, model):

        value = float(value)
        score = value * self.sign

        if self.best_score is None:

            self.best_score = score
            self.save_checkpoint(value, model)

        elif score < self.best_score + self.delta:

            self.counter += 1

            if self.verbose:
                print(
                    f"EarlyStopping counter: "
                    f"{self.counter} out of {self.patience}"
                )

            if self.counter >= self.patience:
                self.early_stop = True

        else:

            self.best_score = score
            self.save_checkpoint(value, model)
            self.counter = 0

    def save_checkpoint(self, value, model):

        if self.verbose:
            print(
                f"Test {self.metric}: "
                f"{self.val_bound:.6f} --> {value:.6f}. "
                "Saving model..."
            )

        torch.save(model.state_dict(), self.save_path)

        self.val_bound = value


# -----------------------------------
# 7. TRAINING SETTINGS
# -----------------------------------

LR = 0.01
WEIGHT_DECAY = 5e-4

TRAIN_ROUNDS = 10
EPOCHS = 1000

PATIENCE = 30
DELTA = 0.0001

HIDDEN_DIM = 16
ATTENTION_HEADS = 4
DROPOUT = 0.5

results = []

print("\nModel: Graph Attention Network (GAT)")
print("Hidden dimension:", HIDDEN_DIM)
print("Attention heads:", ATTENTION_HEADS)
print("Dropout:", DROPOUT)
print("Learning rate:", LR)
print("Training rounds:", TRAIN_ROUNDS)


# -----------------------------------
# 8. TRAINING
# -----------------------------------

for r in trange(TRAIN_ROUNDS):

    print(f"\nTraining round: {r + 1}/{TRAIN_ROUNDS}")

    model = GATNet(
        input_dim=data.num_node_features,
        hidden_dim=HIDDEN_DIM,
        num_classes=2,
        heads=ATTENTION_HEADS,
        dropout=DROPOUT
    ).to(device)

    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=LR,
        weight_decay=WEIGHT_DECAY
    )

    best_model_path = os.path.join(
        MODEL_DIR,
        f"GAT_manual+lstm_seq74_{r}round_best.pt"
    )

    early_stopping = EarlyStopping(
        save_path=best_model_path,
        verbose=True,
        patience=PATIENCE,
        delta=DELTA,
        metric='loss'
    )

    train_loss_s = []
    test_loss_s = []

    for epoch in range(EPOCHS):

        model.train()

        optimizer.zero_grad()

        out = model(data)

        loss = F.nll_loss(
            out[data.train_mask],
            data.y[data.train_mask]
        )

        loss.backward()

        optimizer.step()

        train_loss_s.append(loss.item())

        model.eval()

        with torch.no_grad():

            out = model(data)

            test_loss = F.nll_loss(
                out[data.test_mask],
                data.y[data.test_mask]
            )

            test_loss_value = test_loss.item()

            pred = out[data.test_mask].argmax(dim=1)

            true = data.y[data.test_mask]

            test_acc = accuracy_score(
                true.cpu().numpy(),
                pred.cpu().numpy()
            )

        test_loss_s.append(test_loss_value)

        if (epoch + 1) % 10 == 0:

            print(
                f"Epoch {epoch + 1} | "
                f"Train Loss: {loss.item():.6f} | "
                f"Test Loss: {test_loss_value:.6f} | "
                f"Test Accuracy: {test_acc:.4f}"
            )

        early_stopping(test_loss_value, model)

        if early_stopping.early_stop:

            print("Early stopping at epoch:", epoch + 1)

            break

    # Load the best checkpoint for this round

    model.load_state_dict(
        torch.load(
            best_model_path,
            map_location=device,
            weights_only=True
        )
    )

    model.eval()

    with torch.no_grad():

        out = model(data)

        pred = out[data.test_mask].argmax(dim=1)

        true = data.y[data.test_mask]

        y_pred = pred.cpu().numpy()

        y_true = true.cpu().numpy()

    # Calculate evaluation metrics

    test_acc = accuracy_score(y_true, y_pred)

    test_pre = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )

    test_rec = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )

    test_f1 = f1_score(
        y_true,
        y_pred,
        zero_division=0
    )

    results.append({

        "model": "GAT",

        "features": "manual+lstm_seq74",

        "round": r,

        "test_acc": test_acc,

        "test_pre": test_pre,

        "test_rec": test_rec,

        "test_f1": test_f1,

        "epoch": len(train_loss_s),

        "best_test_loss": early_stopping.val_bound

    })

    print(
        f"\nRound {r + 1} results:"
        f"\nAccuracy: {test_acc:.4f}"
        f"\nPrecision: {test_pre:.4f}"
        f"\nRecall: {test_rec:.4f}"
        f"\nF1-score: {test_f1:.4f}"
    )

    # Plot training and test loss

    plt.figure(figsize=(8, 4))

    plt.plot(train_loss_s, label="Training loss")

    plt.plot(test_loss_s, label="Test loss")

    plt.xlabel("Epoch")

    plt.ylabel("Loss")

    plt.title(f"GAT Round {r + 1}")

    plt.legend()

    plt.tight_layout()

    plt.show()

    del model

    if torch.cuda.is_available():

        torch.cuda.empty_cache()


# -----------------------------------
# 9. SAVE RESULTS
# -----------------------------------

print("\nTraining completed.")

results_df = pd.DataFrame(results)

RESULT_PATH = os.path.join(
    RESULT_DIR,
    "GAT_manual+lstm_seq74_result.csv"
)

results_df.to_csv(
    RESULT_PATH,
    index=False
)

display(results_df)

print("\nAverage performance across 10 rounds:")

for metric in [
    "test_acc",
    "test_pre",
    "test_rec",
    "test_f1"
]:

    print(
        f"{metric}: {results_df[metric].mean():.4f}"
    )

print("\nResults saved to:", RESULT_PATH)


# -----------------------------------
# 10. FINAL ROUND CLASSIFICATION REPORT
# -----------------------------------

final_checkpoint = results[-1]["round"]

best_model_path = os.path.join(
    MODEL_DIR,
    f"GAT_manual+lstm_seq74_{final_checkpoint}round_best.pt"
)

final_model = GATNet(
    input_dim=data.num_node_features,
    hidden_dim=HIDDEN_DIM,
    num_classes=2,
    heads=ATTENTION_HEADS,
    dropout=DROPOUT
).to(device)

final_model.load_state_dict(
    torch.load(
        best_model_path,
        map_location=device,
        weights_only=True
    )
)

final_model.eval()

with torch.no_grad():

    out = final_model(data)

    y_pred = out[data.test_mask].argmax(dim=1).cpu().numpy()

    y_true = data.y[data.test_mask].cpu().numpy()

print("\nClassification Report:")

print(
    classification_report(
        y_true,
        y_pred,
        labels=[0, 1],
        target_names=["Class 0", "Class 1"],
        zero_division=0
    )
)

print("Confusion Matrix:")

print(
    confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    )
)
```
